# Shepp-Logan Phantom Analysis

## Objective

This notebook introduces the representation and analysis of a synthetic
medical imaging phantom.

The Shepp-Logan phantom is used as a controlled reference image for
medical imaging experiments. Because its structure is known, it provides
a useful starting point for studying image formation, reconstruction,
image processing, and image quality.

In this experiment, we will:

- Generate a Shepp-Logan phantom.
- Inspect its numerical representation.
- Analyze its intensity range and basic statistics.
- Visualize the image.
- Study the distribution of pixel intensities.

## 1. Image Representation

A digital medical image can be represented as a two-dimensional numerical
array.

Each element of the array corresponds to a pixel and stores an intensity
value.

For an image with height $H$ and width $W$, the image can be represented as:

$$
I \in \mathbb{R}^{H \times W}
$$

In this project, the phantom is represented as a NumPy array. The array
contains the numerical information that will later be used for image
processing and CT simulation.

```python
import matplotlib.pyplot as plt
import numpy as np

from medimaging.simulation.phantom import create_shepp_logan
```

## 2. Generate the Phantom

We generate a 256 × 256 Shepp-Logan phantom using the function
implemented in the `medimaging.simulation` module.

The phantom is returned as a NumPy array using the `float32` data type.

```python
phantom = create_shepp_logan(256)

print("Shape:", phantom.shape)
print("Dtype:", phantom.dtype)
print("Minimum:", phantom.min())
print("Maximum:", phantom.max())
```

### Observations

The phantom is represented as a square two-dimensional array.

The shape indicates the spatial dimensions of the image, while the data
type determines how the pixel intensities are stored numerically.

The minimum and maximum values describe the intensity range of the image.

## 3. Visualizing the Phantom

The numerical array can be displayed as an image by mapping each pixel
intensity to a grayscale value.

Low intensity values are displayed as darker pixels, while higher
intensities are displayed as brighter pixels.

```python
plt.figure(figsize=(6, 6))

plt.imshow(phantom, cmap="gray")

plt.title("Shepp-Logan Phantom")
plt.axis("off")

plt.show()
```

## 4. Pixel Intensity Distribution

The histogram describes how the pixel intensities are distributed across
the image.

To construct the histogram, the two-dimensional image is temporarily
flattened into a one-dimensional array using `ravel()`.

This allows us to analyze the distribution of all pixel intensities
without considering their spatial location.

```python
plt.figure(figsize=(8, 4))

plt.hist(phantom.ravel(), bins=50)

plt.title("Distribution of Pixel Intensities")
plt.xlabel("Intensity")
plt.ylabel("Number of Pixels")

plt.show()
```

## 5. Image Statistics

Basic statistical measurements provide a quantitative description of
the image.

We calculate:

- Mean intensity.
- Standard deviation.
- Number of non-zero pixels.
- Total number of pixels.

```python
print("Mean:", phantom.mean())
print("Standard deviation:", phantom.std())
print("Non-zero pixels:", np.count_nonzero(phantom))
print("Total pixels:", phantom.size)
```

## 6. Interpretation

The phantom provides a controlled image in which the spatial structure
and intensity distribution are known.

The large number of pixels with low intensity corresponds to regions
outside the synthetic structures represented by the phantom.

The variation in intensity values reflects the different regions defined
within the Shepp-Logan model.

This controlled representation will be useful in later experiments because
we will be able to compare an original reference image with images
obtained after simulated acquisition, noise, and reconstruction.
